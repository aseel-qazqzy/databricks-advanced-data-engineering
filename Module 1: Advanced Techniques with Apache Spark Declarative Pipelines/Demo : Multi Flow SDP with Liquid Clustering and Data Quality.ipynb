### Explore the data inside the volumns 

#### 1- bright_home_orders

```python
spark.sql(f"LIST '{my_vol_path}/bright_home_orders'").display()
```

```python
# a. Row count
df_count = spark.sql(f"""
        SELECT COUNT(*) AS Total_rows 
        FROM read_files('{my_vol_path}/bright_home_orders')         

""")

display(df_count)

# b. Schema
df_schema = spark.sql(f"""
  DESCRIBE SELECT * FROM read_files('{my_vol_path}/bright_home_orders')                       
""")

display(df_schema)

# c.Preview rows 
df_preview = spark.sql(f"""
      SELECT * 
      FROM read_files('{my_vol_path}/bright_home_orders')
      LIMIT 5
""")

display(df_preview)
```
